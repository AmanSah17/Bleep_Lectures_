## 9. Lists

A list is an ordered collection of elements. It is one of the most common built-in data structures in Python.

## What is a list?

A list is an ordered collection of elements.

```python
numbers = [10, 20, 30, 40]

print(numbers)
print(type(numbers))
```

Output:

```python
[10, 20, 30, 40]
<class 'list'>
```

A list can contain different data types:

```python
data = [10, 3.14, "Python", True]
print(data)
```

It can also contain another list:

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix)
```

This becomes important later when discussing matrices, tensors, and NumPy arrays.

---

## Lists are ordered

Every element has an index.

```python

print(numbers[0])
print(numbers[1])

```

Output:

```python

20
40
```

Python uses zero-based indexing.

```python
Index:     0    1    2    3
           ↓    ↓    ↓    ↓
List:     10   20   30   40
```

---

## Lists are mutable

Mutable means we can change the object after creating it.


Output:

```python

```

We can add elements:

```python

print(numbers)
```

Output:

```python
[10, 200, 30, 40]
```

We can remove elements:

```python
numbers.remove(200)
print(numbers)
```

## 9. Tuples

A tuple is an ordered, immutable collection of values.

Unlike a list, once a tuple is created, its elements cannot be changed.

```python
point = (10, 20)
print(point)
print(type(point))
```

Why use tuples?
- fixed data such as coordinates
- safe values that should not change
- returning multiple values from a function



````python
# Tuple: ordered and immutable collection.
# A tuple stores multiple values together, and it cannot be modified once created.
tuple_1 = (11,222,1555,5555)
type(tuple_1)

# Access by index. Indexing is still zero-based like lists.
tuple_1[1]

# gets the very ffirst
tuple_1[0]







````

````python
# Tuple packing: assigning multiple values to one tuple.
student = ("Ahhaysgd", 21, "AI")
print(student)

# Tuple unpacking: split a tuple into separate variables.

student_name , student_age , student_interest = student
````

````python
student_name
````

````python

# Tuples can contain different data types.
profile = ("Aman", 21, True, "Bhopal",21,22,23)
print(profile)

# count(value): returns how many times a value appears.
print(profile.count(21))

# # index(value): returns the position of the first matching item.
print(profile.index(True))

print(profile.index(21))
````

````python
# A common use case: returning multiple values from a function.
def get_student_info():
    return ("Aman", 22, "Machine Learning")

student_name, student_age, student_course = get_student_info()
print(student_name, student_age, student_course)

````

````python
# Tuples are safer for fixed data because they are immutable.
# Example: coordinates or RGB values
rgb = (255, 0, 128)
red, green, blue = rgb
print(red, green, blue)
````

### Tuple methods and when to use them

Tuple methods are limited because the object is immutable, but they are still useful:

- `count(value)`: count how many times a value appears
- `index(value)`: find the first index where a value exists

```python
numbers = (10, 20, 10, 30, 10)

print(numbers.count(10))
print(numbers.index(30))
```

A tuple is helpful when you want values to stay constant, like:
- RGB colors
- database records
- fixed configuration values
- coordinates

```python
color = (255, 255, 0)
print(color)
# Here the color values should not change while the program runs.
```


````python
# Advanced tuple example: unpacking and fixed records.

student_record = ("Shikhar Dhawan ", "AI", 90, 2026)
name, course, score, year = student_record

print(name)
print(course)
print(score)
print(year)






````

````python
# count(value): counts repeated values.
marks = (88, 92, 88, 95, 88)
print(marks.count(88))
````

````python
# index(value): finds the first occurrence of a value.
print(marks.index(95))
````

````python

# A more realistic example: coordinates.
point_a = (3, 4)
point_b = (8, 9)

x1 ,y1 = point_a
x2 ,y2 = point_b



print(x1, y1)
print(x2, y2)
````

## 10. Sets

A set is an unordered collection of unique elements.

It is useful when:
- you want to remove duplicates
- order does not matter
- you need membership checks quickly

```python
numbers = {10, 20, 30, 20, 10}
print(numbers)
```

A set automatically removes duplicates.


````python
# Set: unordered collection with unique values.
# Sets are useful for removing duplicates and checking membership quickly.
items = {"apple" , "banana","Apple" , "apple","kiwi",'orange', "Orange2","orange"}

print(type(items))

items
````

````python
set_numbers = {10,22,222,10,15,15,26,222}
print(set_numbers)
````

````python
# add(value): insert a new element into the set.
items = {"apple" , "banana","Apple" , "apple","kiwi",'orange', "Orange2","orange"}
items.add("grape")
print(items)
````

````python
# remove(value): removes an element; raises error if missing.
# discard(value): removes an element if it exists without error.
#items.remove("mango")
items.discard("mango")
print(items)
````

````python
# pop(): removes and returns an arbitrary element.
print(items.pop())
print(items)
````

````python
# clear(): empties the set.
items.clear()
print(items)
````

````python
# Set operations are very useful in data tasks.
math_students = {"Akaansha", "Riya", "Neha","Mouktik","Bindu","Aashi"}
cs_students = {"Aman", "Karan", "Riya","Mouktik","Bindu","aakash","bhargav"}
````

````python

# union(): combines all unique values from both sets.
print(f"Union of the students",math_students.union(cs_students))

# intersection(): common elements only.
print(f"Intersection", math_students.intersection(cs_students))


# difference(): elements in one set but not the other.
print(f"difference", math_students.difference(cs_students))

# symmetric_difference(): elements in exactly one set.
print(f"symmetric_difference", math_students.symmetric_difference(cs_students))
````

````python
# A practical example: remove duplicate names from a list.
name_list = ["Abhishek", "Aman", "Neha", "Riya", "Aman", "Neha","Shekhar"]
unique_names = set(name_list)   # type conversion list ------ set # set are unique elemts . 
print(unique_names)
print(len(unique_names))
````

## 13. Boolean and None

Boolean values represent truth or falsehood.

- `True` means yes / correct / logically valid
- `False` means no / incorrect / logically invalid
- `None` means no value is assigned

```python
is_active = True
is_logged_in = False
result = None
```

Booleans are used heavily in conditions, filtering, and logic.


````python
print(bool(1)) #1 treated as True

print(bool(0)) #0 treated as False      


print(bool("Python"))



print(bool(""))
print(bool(None))
````

````python

# Booleans are used for conditions in if statements.
score = 80
if score >= 75:
    print("Pass")
else:
    print("Fail")
````

````python
# Logical operations with booleans.
passed_exam = True
has_attendance = False
# eligible = passed_exam and has_attendance
````

````python


# None represents missing data or an uninitialized value.
result = None
print(result)
print(type(result))

# Use is None to check whether a variable has no value.
if result is None:
    print("No result yet")
````

````python

# Advanced concept: comparing values gives boolean results.
print(10 > 5)
print(10 == 10)
print(3 != 5)
````
