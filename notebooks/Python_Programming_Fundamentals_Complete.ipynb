# 2. How Python Programs Execute

When we write:

```python
print("Hello World")
```

the text is not directly executed as CPU machine instructions.

A simplified CPython execution model is:

```text
Python Source Code (.py)
        ↓
Lexing / Parsing / Compilation
        ↓
Python Bytecode
        ↓
Python Virtual Machine
        ↓
Interaction with the operating system / CPU
        ↓
Program Output
```

## Important terminology

### Source code

The human-readable Python program stored in a `.py` file.

### Bytecode

An intermediate instruction representation generated by CPython from Python source code.

### Python Virtual Machine

The execution machinery in CPython that interprets Python bytecode.

### Interpreter

The term is commonly used to describe Python because Python programs are executed through an interpreter/runtime rather than being distributed as native machine-code executables in the usual workflow.

> **Important:** Saying "Python is interpreted" does not mean that Python source is never compiled. CPython compiles source code into bytecode before executing it.

## `.pyc` files

CPython may cache compiled bytecode in `__pycache__` directories using `.pyc` files.

A `.pyc` file is **not** the same thing as a native CPU executable such as a Windows `.exe`.

### Truth-table intuition

For `and`, both operands must be truthy for the result to be truthy.

For `or`, at least one operand must be truthy.

For `not`, the truth value is inverted.

Python's `and` and `or` also use short-circuit evaluation and return operands rather than necessarily returning `True` or `False`.


````python
print(True and "Python")
print(False and "Python")
print(False or "Python")
print("Hello" or "Python")
````

## Identity operators

Use:

- `is`
- `is not`

to test object identity.

Do not use `is` as a replacement for `==`.


````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print("a == b:", a == b)
print("a is b:", a is b)
print("a is c:", a is c)
````

The distinction is:

```text
==  → equality comparison
is  → object identity comparison
```

For example, two different list objects can contain equal values while still being different objects.

# 18. Scope and LEGB

A variable's scope determines where a name can be resolved.

Python's name-resolution rule is commonly summarized as **LEGB**:

```text
L → Local
E → Enclosing
G → Global
B → Built-in
```

Example:


````python
x = "global"

def outer():
    x = "enclosing"

    def inner():
        x = "local"
        print(x)

    inner()

outer()
print(x)
````

## Local versus global names

A local assignment inside a function creates or rebinds a local name unless `global` is used.

Avoid unnecessary global state in normal application code because it makes programs harder to reason about and test.


````python
x = 100

def test():
    x = 50
    print("inside:", x)

test()
print("outside:", x)
````

## Mini Project 1 — Calculator

### Requirements

Build a calculator that:

1. Reads two numbers.
2. Reads an operator.
3. Performs the requested operation.
4. Displays the result.
5. Handles division by zero.

Suggested operators:

```text
+
-
*
/
```


````python
def calculator(a, operator, b):
    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        if b == 0:
            return "Error: division by zero"
        return a / b
    else:
        return "Error: unsupported operator"

print(calculator(10, "+", 5))
print(calculator(10, "/", 2))
print(calculator(10, "/", 0))
````

## Mini Project 2 — Student Grade Calculator

### Requirements

Create a program that accepts marks and calculates:

- Total
- Average
- Grade

Suggested grading logic:

```text
90+  → A
80–89 → B
70–79 → C
60–69 → D
Below 60 → F
```


````python
def calculate_grade(marks):
    average = sum(marks) / len(marks)

    if average >= 90:
        grade = "A"
    elif average >= 80:
        grade = "B"
    elif average >= 70:
        grade = "C"
    elif average >= 60:
        grade = "D"
    else:
        grade = "F"

    return average, grade

marks = [85, 92, 78, 88, 90]

average, grade = calculate_grade(marks)

print("Total:", sum(marks))
print("Average:", average)
print("Grade:", grade)
````
