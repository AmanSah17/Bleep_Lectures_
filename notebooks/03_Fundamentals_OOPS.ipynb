# What is OOP?
Object-Oriented Programming (OOP) is a programming approach in which we organize code around objects that contain both data and functions that operate on that data.
Let's understand this using a familiar example: a student.
A student has data such as:
- Name
- Roll number
- Marks
- Department
A student can also perform actions such as:
- Display details
- Calculate percentage
- Check whether they passed an examination
In OOP, we can combine these related properties and actions into a single logical unit called an object.

## The percentage calculation here is deliberately simple because we're assuming each student has one aggregate score out of 100.
As the program grows, we face several problems:
1. Repetition: We repeatedly create variables and write similar logic.
2. Organization: A student's name, marks, and roll number are scattered across separate variables.
3. Maintenance: If we change how student results are calculated, we must manage that logic consistently.
4. Scalability: Managing thousands of students becomes harder if every student requires separately written code.
5. Reusability: We need a standard structure for every student.
We can improve this with functions and data structures, too. OOP provides another useful option: create a reusable blueprint for each student

# What is __init__()?
__init__() is a special method that Python calls automatically to initialize a newly created object.
It allows us to provide the required information when an object is created.


```python
# Standard way of writing a class # Storing attrbute values

class Student:

    def __init__(self, name, age, roll_number, marks, section):
        self.name = name
        self.age = age
        self.roll_number = roll_number
        self.marks = marks
        self.section = section

        
```
