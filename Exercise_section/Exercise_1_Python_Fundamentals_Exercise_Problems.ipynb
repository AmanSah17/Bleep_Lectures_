## Problem 3: Temperature Conversion

### Problem
A temperature is 30°C. Convert it into Fahrenheit.

### Detailed Explanation
The formula to convert Celsius to Fahrenheit is:

F = (C × 9/5) + 32

Substitute 30 for C:
- F = (30 × 9/5) + 32
- 30 × 9/5 = 54
- 54 + 32 = 86

Therefore, 30°C is equal to 86°F.

### Concept Covered

- Operator precedence
- Type conversion awareness

### Why It Matters
This is a classic real-life problem. It shows how formulas in mathematics become expressions in Python.

## Problem 7: Palindrome Check

### Problem
Check whether the word "level" is a palindrome.

### Detailed Explanation
A palindrome is a word or phrase that reads the same forward and backward.

Take the word "level":
- Forward: l e v e l
- Backward: l e v e l

The characters match exactly when read in reverse order. Therefore, "level" is a palindrome.

### Concept Covered
- String comparison
- Reverse thinking
- Logical pattern checking

### Why It Matters
Palindrome problems teach how to compare strings and reason about positions. They are often used in exams and coding interviews.

```python
string = "Level"
string = string.lower()  # Convert to lowercase for case-insensitive comparison
if string[0] == string[-1]:
    print("initial and last word are same")


```

```python
def is_palidrome(string:str) -> bool:
    cleaned_words = string.lower().replace(" ", "")  # Convert to lowercase and remove spaces
    return cleaned_words == cleaned_words[::-1]


print(is_palidrome("Level"))  # True
print(is_palidrome("Python"))  # False
```

### Function Example

A palindrome reads the same forward and backward. To check this in Python, we can reverse the string and compare it to the original.

Step by step:
1. Take the input word.
2. Reverse it using slicing: `word[::-1]`.
3. Compare the original word with the reversed word.
4. If both are equal, it is a palindrome.

This is a clean way to write the solution as a function.


```python
def is_palindrome(word):
    reversed_word = word[::-1]
    if word == reversed_word:
        return True
    else:
        return False

```

## Problem 8: Count Vowels in a Name

### Problem
Count the number of vowels in the word "education".

### Detailed Explanation
The vowels in English are:
- a, e, i, o, u

Now look at the word "education":
- e, u, a, i, o

Count each vowel:
- e → 1
- u → 1
- a → 1
- i → 1
- o → 1

So there are 5 vowels in the word.

### Concept Covered
- String iteration
- Character checking
- Basic text analysis

### Why It Matters
This kind of problem introduces the idea of scanning characters in a string. It is a foundational skill for text processing and NLP tasks.

### 3 Beginner-Friendly Ways to Solve It

Here are three common methods to count vowels in a word.

1. Use a loop and check each character.
2. Use `count()` for each vowel separately.
3. Use a set and `sum()` with a short condition.

All three methods work for the word "education" and give the same answer: 5.


```python
word = "Python"
vowels = "aeiou"

# Method 1: Loop through each character

def count_vowels_loop(name):
    count = 0
    for ch in name.lower():
        if ch in vowels:
            count += 1
    return count





print("Method 1:", count_vowels_loop(word))

```

```python
word = "Education"
vowels = "aeiou"

# Method 2: Use count() for each vowel

def count_vowels_count(name):
    total = 0
    for v in vowels:
        total += name.lower().count(v)   #count methods
    return total


print("Method 2:", count_vowels_count(word))
```

```python
# Method 3: Use set() and sum()

def count_vowels_set(name):
    return sum(1 for ch in name.lower() if ch in set(vowels))


print("Method 3:", count_vowels_set(word))
```
